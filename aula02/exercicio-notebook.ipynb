```python
import seaborn as sns
```

```python
grafico = sns.histplot(data=df, x="categoria",kde=False,bins=20)
grafico.set_title("Tabela de produtos mais vendidos")
grafico.set_xlabel("Categorias")
grafico.set_ylabel("Quantidade vendida")
```

```python
grafico = sns.boxplot(data=df, x="categoria", y="preco_unitario")
grafico.set_title("Preço por categoria")
grafico.set_xlabel("Categorias")
grafico.set_ylabel("Preços")
```
